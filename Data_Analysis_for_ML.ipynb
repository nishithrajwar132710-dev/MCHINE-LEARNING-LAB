# Data Analysis for Machine Learning
**Aim:** To perform data analysis (EDA) and data preprocessing on a dataset before using it in machine learning.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

### Load data

```python
df = sns.load_dataset('titanic')
df.head()
```

### Basic information

```python
print("Shape:", df.shape)
df.info()
```

```python
df.describe()
```

### Missing values

```python
print(df.isnull().sum())

plt.figure(figsize=(8,4))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values")
plt.show()
```

### Handling missing values

```python
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df = df.drop(['deck', 'embark_town', 'alive', 'class', 'who', 'adult_male', 'alone'], axis=1)

print(df.isnull().sum())
```

### Duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print(df.shape)
```

### Univariate analysis

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
sns.histplot(df['age'], kde=True)
plt.title("Age")

plt.subplot(1,3,2)
sns.countplot(x='survived', data=df)
plt.title("Survived")

plt.subplot(1,3,3)
sns.boxplot(y=df['fare'])
plt.title("Fare (outliers)")

plt.tight_layout()
plt.show()
```

### Bivariate analysis

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Sex vs Survived")

plt.subplot(1,3,2)
sns.countplot(x='pclass', hue='survived', data=df)
plt.title("Class vs Survived")

plt.subplot(1,3,3)
sns.boxplot(x='survived', y='age', data=df)
plt.title("Age vs Survived")

plt.tight_layout()
plt.show()
```

### Outlier removal (fare)

```python
q1 = df['fare'].quantile(0.25)
q3 = df['fare'].quantile(0.75)
iqr = q3 - q1
df = df[(df['fare'] >= q1 - 1.5*iqr) & (df['fare'] <= q3 + 1.5*iqr)]
print("Shape after removing outliers:", df.shape)
```

### Encoding categorical data

```python
df['sex'] = df['sex'].map({'male': 0, 'female': 1})
df = pd.get_dummies(df, columns=['embarked'], drop_first=True)
df.head()
```

### Correlation

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

### Feature scaling

```python
cols = ['age', 'fare']
sc = StandardScaler()
df[cols] = sc.fit_transform(df[cols])
df.head()
```

**Conclusion:** EDA was done on the Titanic dataset. Missing values were filled, duplicates and outliers were removed, categorical columns were encoded and numeric columns were scaled, so the data is now ready for a machine learning model. Sex and passenger class show the strongest relation with survival.
