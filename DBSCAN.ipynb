# DBSCAN Algorithm
**Aim:** To implement DBSCAN (Density Based Spatial Clustering of Applications with Noise) clustering algorithm.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN, KMeans
from sklearn.metrics import silhouette_score
```

### Create data

```python
X, y = make_moons(n_samples=300, noise=0.07, random_state=42)
X = StandardScaler().fit_transform(X)
print(X.shape)

plt.scatter(X[:,0], X[:,1], s=15)
plt.title("Original Data")
plt.show()
```

### Apply DBSCAN

```python
db = DBSCAN(eps=0.3, min_samples=5)
labels = db.fit_predict(X)

n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = list(labels).count(-1)

print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)
```

### Plot clusters

```python
plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis', s=15)
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

### Different values of eps

```python
for e in [0.1, 0.2, 0.3, 0.5]:
    d = DBSCAN(eps=e, min_samples=5)
    l = d.fit_predict(X)
    c = len(set(l)) - (1 if -1 in l else 0)
    print("eps =", e, "clusters =", c, "noise =", list(l).count(-1))
```

### Compare with KMeans

```python
km = KMeans(n_clusters=2, n_init=10, random_state=42)
km_labels = km.fit_predict(X)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1], c=labels, s=15)
plt.title("DBSCAN")
plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1], c=km_labels, s=15)
plt.title("KMeans")
plt.show()
```

**Conclusion:** DBSCAN was implemented on moon shaped data. With eps = 0.3 it found the two clusters correctly, while KMeans could not separate the moon shapes properly. Very small eps (0.1) gave many clusters and noise points, and large eps (0.5) merged everything into one cluster, so the result depends on eps and min_samples.
